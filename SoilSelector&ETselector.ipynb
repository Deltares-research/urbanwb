### Soil selector

```python
path = 'C:/Users/ZWX/PythonNotebooks/UWBM/Trial/'
soilmatrix = pd.read_csv(path + 'soilparameter.csv')
```

```python
soilmatrix.head()
```

```python
soilmatrix.shape  
```

```python
def SoilSelector():
    a = input('Please choose soil type:')
    b = input('Please choose thickness of rootzone [cm]:')
    c = input('Please type in GWL [m -MSL]:')
    c = float(c)
    if c>= 0.0 and c <= 2.5:
        c = float(Decimal(str(c)).quantize(Decimal('.1'), rounding=ROUND_FLOOR))
    elif c < 3.0:
        c = 2.5
    elif c < 5.0:
        c = int(c)
    elif c <= 10:
        c = 5.0
    else:
        c = 10.0
        
    sol = soilmatrix.loc[(soilmatrix.soil_type == int(a)) & (soilmatrix.th_rz == int(b)) & (soilmatrix.gwl == c)]

    return sol
```

```python
SoilSelector()
```

```python
etmatrix = pd.read_csv(path + 'ETparameter.csv')

def ETSelector():
    a = input('Please choose soil type:')
    b = input('Please choose crop type:')
    
    sol = etmatrix.loc[(etmatrix.soil_type == int(a)) & (etmatrix.crop_type == int(b))]
    
    return sol
```
